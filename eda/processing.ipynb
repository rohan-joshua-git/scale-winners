```python
dt_df = df.with_columns([
    pl.col(c).str.to_datetime(strict=False) for c in ["INITIATED_AT", "SETTLED_AT", "CREATED_AT", "UPDATED_AT"]
] + [
    pl.col("VALUE_DATE").str.to_date(strict=False)
])
```

```python
dt_df.head()
```

```python
import numpy as np
```

```python


dt_df = dt_df.with_columns([
    pl.col("INITIATED_AT").dt.hour().alias("hour"),
    pl.col("INITIATED_AT").dt.weekday().alias("weekday"),  # 1=Mon..7=Sun
])


```

```python
feature_df = dt_df.with_columns([
    (2 * np.pi * pl.col("hour") / 24).sin().alias("hour_sin"),
    (2 * np.pi * pl.col("hour") / 24).cos().alias("hour_cos"),
    (2 * np.pi * pl.col("weekday") / 7).sin().alias("weekday_sin"),
    (2 * np.pi * pl.col("weekday") / 7).cos().alias("weekday_cos"),
])
```

```python
feature_df.head()
```

```python
import polars as pl
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# 1. Inspect what you're working with
print(feature_df.head())
print(feature_df.dtypes)

# 2. Separate the label from the features
# PRIORITY_FLAG is presumably categorical (e.g. "HIGH"/"LOW"/"MEDIUM")
# Isolation Forest is unsupervised, so we only use it to evaluate AFTER fitting — never as a training input
label_col = "PRIORITY_FLAG"

# Convert label to binary: HIGH = anomaly-like (1), everything else = normal (0)
# Adjust this mapping if PRIORITY_FLAG has different levels
y_true = feature_df.select(
    (pl.col(label_col) == "HIGH").cast(pl.Int8).alias("y_true")
).to_series().to_numpy()

# 3. Build feature matrix - drop label and any obvious ID/string columns
exclude_cols = [label_col]
non_numeric_cols = [c for c, dt in zip(feature_df.columns, feature_df.dtypes) 
                     if dt not in (pl.Float32, pl.Float64, pl.Int8, pl.Int16, pl.Int32, pl.Int64)]

feature_cols = [c for c in feature_df.columns if c not in exclude_cols and c not in non_numeric_cols]
print(f"\nUsing {len(feature_cols)} numeric features: {feature_cols}")

X = feature_df.select(feature_cols).to_numpy()

# Check for NaNs before fitting - Isolation Forest will error out otherwise
if np.isnan(X).any():
    print("⚠️  NaNs detected in feature matrix — imputing with column median")
    col_medians = np.nanmedian(X, axis=0)
    inds = np.where(np.isnan(X))
    X[inds] = np.take(col_medians, inds[1])

# 4. Fit naive Isolation Forest (no tuning, defaults + a seed)
iso = IsolationForest(n_estimators=100, contamination="auto", random_state=42)
iso.fit(X)

# -1 = anomaly, 1 = normal -> convert to 1 = anomaly, 0 = normal to match y_true convention
raw_pred = iso.predict(X)
y_pred = (raw_pred == -1).astype(int)
scores = iso.decision_function(X)  # lower = more anomalous

# 5. Compare against PRIORITY_FLAG label
print("\n" + "="*60)
print("Isolation Forest vs PRIORITY_FLAG comparison")
print("="*60)

print(f"\nActual HIGH priority count: {y_true.sum():,} / {len(y_true):,} ({y_true.mean()*100:.2f}%)")
print(f"IF flagged anomaly count:   {y_pred.sum():,} / {len(y_pred):,} ({y_pred.mean()*100:.2f}%)")

print("\nConfusion Matrix (rows=actual, cols=predicted):")
print(confusion_matrix(y_true, y_pred))

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=["NOT_HIGH", "HIGH"]))

try:
    auc = roc_auc_score(y_true, -scores)  # negate: lower decision_function = more anomalous = higher risk
    print(f"ROC-AUC (anomaly score vs HIGH priority label): {auc:.4f}")
except ValueError as e:
    print(f"Could not compute AUC: {e}")

# 6. Attach results back to the dataframe for inspection
feature_df = feature_df.with_columns([
    pl.Series("if_anomaly_score", scores),
    pl.Series("if_is_anomaly", y_pred)
])

# Look at rows where IF disagrees with the label — most interesting cases
disagreements = feature_df.filter(
    (pl.col("if_is_anomaly") == 1) & (pl.col(label_col) != "HIGH")
)
print(f"\nIF flagged as anomaly but NOT labeled HIGH: {disagreements.height:,} rows")

missed = feature_df.filter(
    (pl.col("if_is_anomaly") == 0) & (pl.col(label_col) == "HIGH")
)
print(f"Labeled HIGH but IF did NOT flag: {missed.height:,} rows")
```

```python
true_rate = y_true.mean()
print(f"True HIGH rate: {true_rate:.4f}")

iso = IsolationForest(n_estimators=100, contamination=true_rate, random_state=42)
iso.fit(X)

raw_pred = iso.predict(X)
y_pred = (raw_pred == -1).astype(int)
scores = iso.decision_function(X)

print(confusion_matrix(y_true, y_pred))
print(classification_report(y_true, y_pred, target_names=["NOT_HIGH", "HIGH"]))
```

```python
feature_df = feature_df.with_columns(pl.Series("if_anomaly_score", scores))

k = int(y_true.sum())  # top-1402 most anomalous
top_k = feature_df.sort("if_anomaly_score").head(k)

matched = top_k.filter(pl.col("PRIORITY_FLAG") == "HIGH").height
print(f"Of top {k} most anomalous rows, {matched} are actually labeled HIGH ({matched/k*100:.1f}% precision@k)")
```
